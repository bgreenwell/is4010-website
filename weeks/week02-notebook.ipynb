## The CPTF prompt framework

- **Context:** What repository, file, audience, constraints, and existing behavior matter?
- **Persona:** What useful perspective should the assistant take?
- **Task:** What observable result do you need?
- **Format:** What should the response contain, and what should it omit?

A precise prompt does not guarantee correct code. It makes the result easier to evaluate.
